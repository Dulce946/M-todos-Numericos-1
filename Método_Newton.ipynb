## Formulación del Problema
El **Método de Newton-Raphson** es un procedimiento iterativo basado en el desarrollo en serie de Taylor de una fracción $f(x)$ al rededor de una aproximación inicial $p_0$. Su objetivo es encontrar una raíz de la eciación $f(x) = 0$.

La fórmula recursiva del algoritmo está dada por:
$$p_n = p_{n-1} - \frac{f(p_{n-1})}{f'(p_{n-1})}$$

**Criterios de patrada:**

El algoritmo se detiene de forma exitosa si se satisface la condición de tolerancia:
$$|p_n - p_{n-1}| < TOL$$

O falla de forma controlada si el número de iteraciones excede un límite máximo establecido $N_0$.
